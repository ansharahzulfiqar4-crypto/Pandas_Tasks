```python
#13:Deleting Rows and Columns
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [20, 21, 19, 22],
    "Marks": [45, 90, 75, 40]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

df = df.drop("Age", axis=1)

print("\nAfter deleting Age column:")
print(df)

df = df.drop(0)

print("\nAfter deleting row with index 0:")
print(df)
```

```python
#14: Handling Missing Values
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nMissing Values:")
print(df.isnull())

print("\nCount of Missing Values:")
print(df.isnull().sum())

df_drop = df.dropna()

print("\nAfter dropna():")
print(df_drop)

df_zero = df.copy()
df_zero["Marks"] = df_zero["Marks"].fillna(0)

print("\nAfter filling missing Marks with 0:")
print(df_zero)

df_mean = df.copy()
df_mean["Marks"] = df_mean["Marks"].fillna(df_mean["Marks"].mean())

print("\nAfter filling missing Marks with average:")
print(df_mean)
```

```python
#15: Handling Duplicate Data
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ali"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nDuplicate Rows:")
print(df.duplicated())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(df)
```

```python
#16:Sorting Data
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [70, 95, 80]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nSorted from Lowest to Highest:")
print(df.sort_values("Marks"))

print("\nSorted from Highest to Lowest:")
print(df.sort_values("Marks", ascending=False))
```

```python
#17:Aggregation and GroupBy
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Course": ["AI", "Web", "AI", "Web"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

marks = df["Marks"]

print("\nSum of Marks:")
print(marks.sum())

print("\nAverage Marks:")
print(marks.mean())

print("\nMaximum Marks:")
print(marks.max())

print("\nMinimum Marks:")
print(marks.min())

print("\nNumber of Students:")
print(marks.count())

result = df.groupby("Course")["Marks"].mean()

print("\nAverage Marks for Each Course:")
print(result)
```

```python
#18. Combining DataFrames
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})

final_df = pd.concat([df1, df2], ignore_index=True)

print("After concat():")
print(final_df)

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})

merged_df = pd.merge(students, marks, on="ID")

print("\nAfter merge():")
print(merged_df)
```
